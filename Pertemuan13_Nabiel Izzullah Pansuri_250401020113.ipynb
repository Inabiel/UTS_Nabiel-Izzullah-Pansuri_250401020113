```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons

x, y = make_moons(n_samples=300, noise=0.2, random_state=42)

plt.scatter(x[:,0], x[:,1], c=y, cmap='coolwarm', alpha=0.7)
plt.title('Dataset Non Linear (2 Kelas Bulan Sabit)')
plt.show()
```

```python
# %pip install tensorflow

from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

x_tr, x_te, y_tr, y_te = train_test_split(x, y, test_size=0.2, random_state=42)

model = Sequential([
    Dense(16, activation='relu', input_shape=(2,)),
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy',metrics=['accuracy'])
model.summary()
```

```python
history = model.fit(x_tr, y_tr, epochs=200, validation_split=0.2, verbose=1)
```

```python
from sklearn.preprocessing import StandardScaler
loss, acc = model.evaluate(x_te, y_te, verbose=0)
print(f'Akurasi pada data uji: {acc:.3f}')
plt.plot(history.history['accuracy'], label='Training')

plt.plot(history.history['val_accuracy'], label='Validasi')
plt.xlabel('Epoch'); plt.ylabel('Akurasi')
plt.legend(); plt.title('Kurva Pembelajaran Model')
plt.show()
```

Berdasarkan grafik kurva pembelajaran model, didapatkan hasil bahwa akurasi training validasi saling mendekati, yang menunjukkan hasil yang baik. tidak ada indikasi overfitting dari grafik dikarenakan akurasi training masih mengejar akurasi validasi, dan tidak ada indikasi akurasi validasi turun. 

Dari percobaan ini, saya juga mendapatkan hasil bahwa semakin banyak epoch, kecenderungan akurasi akan naik. namun jika epochnya terlalu banyak, akan menjadi overfitting karena model akan lebih mengenali data training dan menjadi bias.
